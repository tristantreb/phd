#### Create two IA priors, one when AR > 40%, one when AR <= 40%

```python
ar_threshold = 40

df_ia_prior = df.copy()
df_ia_prior[f"AR > {ar_threshold}"] = df_ia_prior["AR mean"] > ar_threshold

# Print number of values above threshold
print(
    f"{df_ia_prior[f'AR > {ar_threshold}'].sum()}/{len(df_ia_prior)} AR values above {ar_threshold}%"
)

ia_dist_high_ar = np.mean(
    df_ia_prior[df_ia_prior[f"AR > {ar_threshold}"]].IA.to_numpy(), axis=0
)
ia_dist_low_ar = np.mean(
    df_ia_prior[~df_ia_prior[f"AR > {ar_threshold}"]].IA.to_numpy(), axis=0
)
```

```python
title = f"Comparison of the IA priors above, below {ar_threshold}% AR learnt on Breathe"

fig = make_subplots(rows=2, cols=1, shared_xaxes=True)
ih.plot_histogram(fig, IA, ia_dist_low_ar, IA.a, IA.b, 1, 1, False)
# name trace to breathe data
fig.data[0].name = f"<= {ar_threshold}% AR"
ih.plot_histogram(fig, IA, ia_dist_high_ar, IA.a, IA.b, 2, 1)
fig.data[1].name = f"> {ar_threshold}% AR"
fig.update_layout(showlegend=True, font=dict(size=10), title=title)
# fig.write_image(dh.get_path_to_main() + f"PlotsCrossStudies/{title}.pdf")
fig.show()
```

```python
# Compute the CPD, IA on the rows, AR on the columns
# cpd = np.zeros((AR.n, IA.n))

cpd_ar_ia = np.zeros((len(IA.bins), len(AR.bins)))
for i in range(len(AR.bins)):
    if i <= AR.get_bin_for_value(40)[1]:
        cpd_ar_ia[:, i] = ia_dist_low_ar
    else:
        cpd_ar_ia[:, i] = ia_dist_high_ar
np.sum(cpd_ar_ia, axis=0)
```

# Generate the breathe data-set using ancestral sampling

Ancestral sampling is a common technique used to generate samples from a joint probability distribution. In Bayesian networks the process begins by sampling from the variables which have no parent using their prior distributions, then sampling the child variables conditioned on these values, and successively doing so by following the direction of the graph's edges. Thus, each variable is sampled given the value its parents in the graph, except the first parents which are sampled from their prior distributions. The term ancestral designates this concept where variables upstream in the graph are considered as ancestors, and determine the value of their descendants. It is particularly useful to generate synthetic data given a probabilistic model.

```python
# Create a dataframe to store the results
df_gen = pd.DataFrame(
    columns=[
        "ID",
        "HFEV1",
        "ecFEV1",
        "AR",
        "HO2Sat",
        "IA",
        "O2Sat",
    ]
)


def generate_synthetic_data_for_ID(df, df_gen):
    """
    Uses ancestral sampling to generate, for each real data-point, a synthetic data-point given the individuals' height, age, sex

    df: the breathe data
    df_gen: the dataframe to store the generated results
    """
    id = df.ID.iloc[0]
    n_datapoints = len(df)

    height = df.Height.iloc[0]
    age = df.Age.iloc[0]
    sex = df.Sex.iloc[0]

    (
        _,
        inf_alg,
        HFEV1,
        ecFEV1,
        AR,
        HO2Sat,
        _,
        IA,
        _,
        O2Sat,
        ) = mb.o2sat_fev1_point_in_time_model_cf_priors(
            height, age, sex, ar_dist_BR.reshape(-1, 1), ia_dist_BR.reshape(-1, 1)
        )

    for _ in range(n_datapoints):
        hfev1_obs = HFEV1.sample()[0]
        ho2sat_obs = HO2Sat.sample()[0]
        ar_obs = AR.sample()[0]
        ia_obs = IA.sample()[0]

        res = ih.infer(
            inf_alg,
            [ecFEV1, O2Sat],
            [[HFEV1, hfev1_obs], [HO2Sat, ho2sat_obs], [AR, ar_obs], [IA, ia_obs]],
        )

        ecfev1_val = ecFEV1.sample(n=1, p=res["ecFEV1 (L)"].values)[0]
        o2sat_val = O2Sat.sample(n=1, p=res["O2 saturation (%)"].values)[0]

        # Populate the results dataframe
        new_row = pd.DataFrame(
            (
                {
                    "ID": id,
                    "HFEV1": hfev1_obs,
                    "ecFEV1": ecfev1_val,
                    "AR": ar_obs,
                    "HO2Sat": ho2sat_obs,
                    "IA": ia_obs,
                    "O2Sat": o2sat_val,
                },
            )
        )
        df_gen = pd.concat([df_gen, new_row], ignore_index=True)

    return df_gen


def generate_synthetic_data_for_ID_ar_ia_factor(df, df_gen):
    """
    Uses ancestral sampling to generate, for each real data-point, a synthetic data-point given the individuals' height, age, sex

    df: the breathe data
    df_gen: the dataframe to store the generated results

    In this function, IA has a parent
    """
    id = df.ID.iloc[0]
    n_datapoints = len(df)

    height = df.Height.iloc[0]
    age = df.Age.iloc[0]
    sex = df.Sex.iloc[0]

    (
        _,
        inf_alg,
        HFEV1,
        ecFEV1,
        AR,
        HO2Sat,
        _,
        IA,
        _,
        O2Sat,
        # ) = mb.o2sat_fev1_point_in_time_model_cf_priors(
        #     height, age, sex, ar_dist_BR.reshape(-1, 1), ia_dist_BR.reshape(-1, 1)
        # )
    ) = mb.o2sat_fev1_point_in_time_model_cf_priors_2(
        height, age, sex, ar_dist_BR.reshape(-1, 1), cpd_ar_ia
    )

    for _ in range(n_datapoints):
        hfev1_obs = HFEV1.sample()[0]
        ho2sat_obs = HO2Sat.sample()[0]
        ar_obs = AR.sample()[0]

        res = ih.infer(
            inf_alg,
            [ecFEV1, O2Sat, IA],
            [[HFEV1, hfev1_obs], [HO2Sat, ho2sat_obs], [AR, ar_obs]],
        )

        ecfev1_val = ecFEV1.sample(n=1, p=res["ecFEV1 (L)"].values)[0]
        o2sat_val = O2Sat.sample(n=1, p=res["O2 saturation (%)"].values)[0]
        ia_val = IA.sample(n=1, p=res["Inactive alveoli (%)"].values)[0]

        # Populate the results dataframe
        new_row = pd.DataFrame(
            (
                {
                    "ID": id,
                    "HFEV1": hfev1_obs,
                    "ecFEV1": ecfev1_val,
                    "AR": ar_obs,
                    "HO2Sat": ho2sat_obs,
                    "IA": ia_val,
                    "O2Sat": o2sat_val,
                },
            )
        )
        df_gen = pd.concat([df_gen, new_row], ignore_index=True)

    return df_gen
```

```python
df_syn = df.groupby("ID").apply(generate_synthetic_data_for_ID, df_gen)
df_syn_ar_ia_factor = df.groupby("ID").apply(generate_synthetic_data_for_ID_ar_ia_factor, df_gen)
```

```python
# Save this dataframe to a file
df_synthetic = df_syn_ar_ia_factor
df_synthetic = df_synthetic.drop(columns=["ID"]).reset_index().drop(columns=["level_1"])
df_synthetic["ecFEV1%"] = df_synthetic["ecFEV1"] / df_synthetic["HFEV1"] * 100
df_synthetic["O2Sat%"] = df_synthetic["O2Sat"] / df_synthetic["HO2Sat"] * 100
df_synthetic["Predicted FEV1"] = df["Predicted FEV1"]
df_synthetic["ecFEV1 % Predicted"] = (
    df_synthetic["ecFEV1"] / df_synthetic["Predicted FEV1"] * 100
)
df_synthetic["Healthy O2 Saturation"] = df["Healthy O2 Saturation"]
df_synthetic["O2 Saturation % Healthy"] = (
    df_synthetic["O2Sat"] / df_synthetic["Healthy O2 Saturation"] * 100
)
df_synthetic.head()
```
